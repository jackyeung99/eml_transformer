```python
FROM_DATE = datetime(2026, 8, 1, tzinfo=timezone.utc)
TO_DATE = datetime(2026, 8, 2, tzinfo=timezone.utc)
```

```python
source = EIA930RegionSource()

records = list(
    source.fetch_records(
        from_date=FROM_DATE,
        to_date=TO_DATE,
    )
)

print("Bronze records:", len(records))

if records:
    pprint.pprint(records[0])
```

```python
rows = []
for i in list(records): 
    record = source.standardize_record(i)
    rows.append(record)

region_df = pd.DataFrame(rows)
```
